4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
# a) factorial function with recursion 
def f(n):
    if n == 1:
        return 1
    for i in range(1,n+1):
        return n * f(n-1)
f(6)
````

````python
# b) factorial without using recursion
def factorial(x):
    a = 1
    for i in range(1,x+1):
        a *= i
    return a
    
factorial(6)
````

````python
from math import prod
def fac(k):
    return prod(range(1 , k+1))

fac(6)
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb":[15,20,30]}
r = [1,2,3]
from math import pi

material = list(densities.keys())
dens = list(densities.values())

for i,j in zip(material, dens):
    for k in r:
        print("Material", i, " with radius=", k, "circle",2*pi*j[0]*k,'\n')
        print("Material", i, " with radius=", k, "disk",pi*j[1]*(k**2),'\n')
        print("Material", i, " with radius=", k, "sphere",(4/3)*j[2]*pi*(k**3),'\n')
````

````python
from math import pi

# Given data
densities = {"Al": [0.5, 1, 2], "Fe": [3, 4, 5], "Pb": [15, 20, 30]}
radii = [1, 2, 3]

# Define the lambda functions for circumference, area, and volume
geo = [
    lambda r: 2 * pi * r,             # circumference for circle
    lambda r: pi * r**2,              # area for disk
    lambda r: (4/3) * pi * r**3       # volume for sphere
]

# Convert densities into a list and use zip to loop through them
material_names = list(densities.keys())
material_densities = list(densities.values())  # Al's densities, Fe's, Pb's

# Compute the weights for each radius and material
for material, dens in zip(material_names, material_densities):
    print(f"\nMaterial: {material}")
    for r in radii:
        weights = [d * f(r) for d, f in zip(dens, geo)]  # zip densities with the geo functions
        print(f"r={r}: Circle={weights[0]:.2f}, Disk={weights[1]:.2f}, Sphere={weights[2]:.2f}")
````

````python
from math import pi
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb":[15,20,30]}
r = [1,2,3]
v = list(densities.values())  # v[0]=Al [linear,surface,volume], v[1]=Fe, v[2]=Pb

for x, y, z in zip(v[0], v[1], v[2]):             # (Al_k, Fe_k, Pb_k) for k=0..2
    for i in range(3):                             # radii r[i]
        if (x, y, z) == (v[0][0], v[1][0], v[2][0]):     # linear → circle
            f = lambda: 2*pi*r[i]*x   # Al
            g = lambda: 2*pi*r[i]*y   # Fe
            w = lambda: 2*pi*r[i]*z   # Pb
            print(f"r:{r[i]}, linear  Al:{f():.2f}, Fe:{g():.2f}, Pb:{w():.2f}")

        elif (x, y, z) == (v[0][1], v[1][1], v[2][1]):   # surface → disk
            f = lambda: pi*(r[i]**2)*x   # Al
            g = lambda: pi*(r[i]**2)*y   # Fe
            w = lambda: pi*(r[i]**2)*z   # Pb
            print(f"r:{r[i]}, surface Al:{f():.2f}, Fe:{g():.2f}, Pb:{w():.2f}")

        elif (x, y, z) == (v[0][2], v[1][2], v[2][2]):   # volumetric → sphere
            f = lambda: (4/3)*pi*(r[i]**3)*x  # Al
            g = lambda: (4/3)*pi*(r[i]**3)*y  # Fe
            w = lambda: (4/3)*pi*(r[i]**3)*z  # Pb
            print(f"r:{r[i]}, volume  Al:{f():.2f}, Fe:{g():.2f}, Pb:{w():.2f}")
    print()
````

````python
# Parent class
class Dog:

    # Class attribute
    species = 'mammal'

    # Initializer / Instance attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_hungry = True

    def eat(self):
        self.is_hungry = False
        

    # instance method
    def description(self):
        return "{} is {} years old".format(self.name, self.age)

    
````

````python
Dogs = [Dog("Tom", 6), Dog("Fletcher", 7), Dog("Larry", 9)]

print(f"I have {len(Dogs)} dogs.")
for i in range(len(Dogs)):
    print(f"{Dogs[i].description()}")
    Dogs[i].eat()
print(f"And they're all {Dog.species},of course.")

if all(dog.is_hungry for dog in Dogs):
    print('My dogs are hungry')
else:
    print('My dogs are not hungry')
````
